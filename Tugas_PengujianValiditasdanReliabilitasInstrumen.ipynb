```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files

print("Library berhasil diimpor.")
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 6

df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

# Mengambil hanya kolom pertanyaan Likert
butir = butir.loc[:, butir.columns.str.contains(r"\[")]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))

display(butir.head())
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

num = butir.replace(skala).apply(
    pd.to_numeric, errors="coerce"
)

# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:",
      int(num.isna().sum().sum()))

display(num.head())
```

```python
dimensi = {}

for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1)
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")

print("\nJumlah dimensi:", len(dimensi))
print("Total butir:", sum(len(daftar) for daftar in dimensi.values()))
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)

print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")
```

```python
hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(
            x[kolom], total - x[kolom]
        )

        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)

display(tabel_validitas)
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (
        1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1)
    )

def kategori(a):
    return (
        "Sangat baik" if a >= 0.90 else
        "Baik" if a >= 0.80 else
        "Reliabel" if a >= 0.70 else
        "Tidak reliabel"
    )

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)

    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a),
    })

    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(
                    cronbach_alpha(x.drop(columns=kolom)), 3
                ),
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus,
    on=["Dimensi", "Butir"],
    how="left"
)

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(
        w,
        sheet_name="Validitas",
        index=False
    )

    tabel_reliabilitas.to_excel(
        w,
        sheet_name="Reliabilitas",
        index=False
    )

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

```python
# Latihan 1: Interpretasi Validitas
tidak_valid = tabel_validitas[
    tabel_validitas["Keterangan"] == "Tidak valid"
]

print("Nilai r tabel :", round(r_tabel, 3))
print("Jumlah butir tidak valid:", len(tidak_valid))

if len(tidak_valid) == 0:
    print("Semua butir dinyatakan valid.")
else:
    display(tidak_valid)

print("\nLima butir dengan r hitung terendah:")
display(
    tabel_validitas.sort_values("r hitung").head(5)
)
```

```python
# Latihan 2: Perbandingan Kriteria Validitas
valid_r = (tabel_validitas["r hitung"] > r_tabel).sum()
valid_corrected = (tabel_validitas["Corrected r"] >= 0.30).sum()

print("Total butir:", len(tabel_validitas))
print("Valid berdasarkan r hitung:", valid_r)
print("Valid berdasarkan corrected r:", valid_corrected)
print("Selisih jumlah:", abs(valid_r - valid_corrected))

print("\nButir dengan corrected r di bawah 0,30:")
display(
    tabel_validitas[
        tabel_validitas["Corrected r"] < 0.30
    ]
)
```

```python
# Latihan 3: Perbaikan Instrumen

nama = list(dimensi.keys())[0]
daftar = dimensi[nama]

x = num[daftar]

alpha_sebelum = cronbach_alpha(x)

kolom_dihapus = daftar[3]

x_baru = x.drop(columns=kolom_dihapus)

alpha_sesudah = cronbach_alpha(x_baru)

print("Dimensi:", nama)
print("Butir yang diuji: B4")
print("Jumlah butir sebelum:", x.shape[1])
print("Jumlah butir sesudah:", x_baru.shape[1])

print("\nAlpha sebelum:", round(alpha_sebelum, 3))
print("Alpha sesudah:", round(alpha_sesudah, 3))
print("Selisih alpha:", round(alpha_sesudah - alpha_sebelum, 3))
```

```python
# Latihan 4: Pengaruh Ukuran Sampel

subset = num.sample(20, random_state=1)

n_subset = len(subset)
df_subset = n_subset - 2
t_subset = stats.t.ppf(0.975, df_subset)

r_tabel_subset = t_subset / np.sqrt(
    df_subset + t_subset**2
)

hasil_subset = []

for nama, daftar in dimensi.items():
    x = subset[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)

        hasil_subset.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Keterangan": (
                "Valid" if r > r_tabel_subset
                else "Tidak valid"
            )
        })

tabel_subset = pd.DataFrame(hasil_subset)

valid_38 = (tabel_validitas["Keterangan"] == "Valid").sum()
valid_20 = (tabel_subset["Keterangan"] == "Valid").sum()

print("PERBANDINGAN UKURAN SAMPEL")
print("--------------------------")
print(f"n = 38 | r tabel = {r_tabel:.3f} | Valid = {valid_38}")
print(f"n = 20 | r tabel = {r_tabel_subset:.3f} | Valid = {valid_20}")

print("\nHasil validitas untuk 20 responden:")
display(tabel_subset)
```

```python
hasil_deskriptif = []
for nama, daftar in dimensi.items():
    skor = num[daftar].mean(axis=1)
    hasil_deskriptif.append({
        "Dimensi": nama.strip(),
        "Rata-rata": round(skor.mean(), 3),
        "Varians": round(skor.var(ddof=1), 3),
        "Standar deviasi": round(skor.std(ddof=1), 3),
        "Skewness": round(skor.skew(), 3),
        "Kurtosis": round(skor.kurt(), 3)
    })

tabel_deskriptif = pd.DataFrame(hasil_deskriptif)
display(tabel_deskriptif)
```
